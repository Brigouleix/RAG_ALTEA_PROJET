## Étape 1 : collecte. Lire les PDF et texte de /docs


```python
from collections import Counter
from pathlib import Path
from pypdf import PdfReader
import json

DOCS_DIR = Path("../docs")   
print(DOCS_DIR.resolve(), "| existe :", DOCS_DIR.exists())
```

```python
def collecter(docs_dir: Path = DOCS_DIR) -> list[dict]:
    """
    Renvoie une liste de pages : {"texte", "source", "page", "service"}.
    
    """
    pages = []
    for fichier in sorted(Path(docs_dir).rglob("*")):
        extension = fichier.suffix.lower()
        if extension not in {".pdf", ".txt"}:
            continue

        service = fichier.parent.name   

        if extension == ".pdf":
            lecteur = PdfReader(fichier)
            for numero, page in enumerate(lecteur.pages, start=1):
                pages.append({
                    "texte": page.extract_text() or "",
                    "source": fichier.name,
                    "page": numero,
                    "service": service,
                })
        else: 
            pages.append({
                "texte": fichier.read_text(encoding="utf-8-sig"),
                "source": fichier.name,
                "page": 1,
                "service": service,
            })
    return pages
```

## Code : Q1, nombre de pages par document

```python
pages = collecter()

par_document = Counter((p["service"], p["source"]) for p in pages)
for (service, source), nb in par_document.items():
    print(f"{service:8} | {source:30} | {nb} page(s)")

print(f"\n{len(par_document)} documents, {len(pages)} pages au total")
print("Pages sans texte :", sum(1 for p in pages if not p["texte"].strip()))
```

```python
Path("../data").mkdir(exist_ok=True)
with open("../data/pages_brutes.json", "w", encoding="utf-8") as f:
    json.dump(pages, f, ensure_ascii=False, indent=2)
print("Sauvegardé :", len(pages), "pages")
```
